```python
# Use pandas to read the tracks_features.csv file
import pandas as pd

# Read the tracks_features.csv file
tracks_features = pd.read_csv('tracks_features.csv')

```

```python
tracks_features.head()
```

```python
# SpotifyManager
from scripts.collect_data.spotify import SpotifyManager

# Initialize the SpotifyManager
spotify_manager = SpotifyManager()

song_ids = song_ids_with_no_genres # tracks_features['id'].tolist()
song_id_to_genres = {}
# Get the genres for the tracks in batches of 100
batch_size = 100
for i in range(0, len(tracks_features), batch_size):
    song_ids_sample = song_ids[i:i+batch_size]
    results = spotify_manager.get_genres_for_songs_batch(song_ids_sample)
    song_ids_to_genres = results[0]
    for song_id, genres in song_ids_to_genres.items():
        song_id_to_genres[song_id] = genres
```

```python
import json
# Write this to a json file
with open('song_id_to_genres.json', 'w') as f:
    json.dump(song_id_to_genres, f)
```

```python
song_ids_with_no_genres = [song_id for song_id, genres in song_id_to_genres.items() if len(genres) == 0]
```
